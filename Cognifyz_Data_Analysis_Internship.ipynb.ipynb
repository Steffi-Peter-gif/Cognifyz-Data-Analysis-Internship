```python
import pandas as pd                                   #level1:task1

# Load the dataset
df = pd.read_csv("Dataset .csv")

# Check basic information
print("Dataset Shape:", df.shape)

# View first few cuisine entries
print("\nSample Cuisine Data:")
print(df['Cuisines'].head())

# Split multiple cuisines and count occurrences
all_cuisines = df['Cuisines'].str.split(', ').explode()

# Find top 3 cuisines
top_cuisines = all_cuisines.value_counts().head(3)

print("\nTop 3 Most Common Cuisines:")
print(top_cuisines)

# Calculate percentages
total_restaurants = len(df)

print("\nPercentage of Restaurants Serving Top Cuisines:")

for cuisine, count in top_cuisines.items():
    percentage = (count / total_restaurants) * 100
    print(f"{cuisine}: {percentage:.2f}%")
```

```python
import matplotlib.pyplot as plt

top_cuisines.plot(kind='bar')

plt.title('Top 3 Most Common Cuisines')
plt.xlabel('Cuisine')
plt.ylabel('Number of Restaurants')

plt.savefig('level1_task1_top_cuisines.png')

plt.show()
```

```python
# Task 2: City Analysis                                        #level1:task2

# 1. Number of restaurants in each city
city_counts = df['City'].value_counts()

print("=== Top 10 Cities by Restaurant Count ===")
print(city_counts.head(10))

# City with highest number of restaurants
top_city = city_counts.idxmax()
top_city_count = city_counts.max()

print("\nCity with Highest Number of Restaurants:")
print(f"{top_city} ({top_city_count} restaurants)")


# 2. Average rating for each city
city_ratings = df.groupby('City')['Aggregate rating'].mean()

print("\n=== Average Rating by City (Top 10) ===")
print(city_ratings.sort_values(ascending=False).head(10))


# 3. City with highest average rating
highest_rating_city = city_ratings.idxmax()
highest_rating = city_ratings.max()

print("\nCity with Highest Average Rating:")
print(f"{highest_rating_city} ({highest_rating:.2f})")
```

```python
import matplotlib.pyplot as plt

city_counts.head(10).plot(kind='bar', figsize=(10,5))

plt.title('Top 10 Cities by Number of Restaurants')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig('level1_task2_city_analysis_chart.png')

plt.show()
```

```python
# Task 3: Price Range Distribution                          #level1:task3

price_counts = df['Price range'].value_counts().sort_index()

print("Price Range Distribution:")
print(price_counts)

# Percentage distribution
price_percentage = (price_counts / len(df)) * 100

print("\nPercentage Distribution:")
print(price_percentage.round(2))
```

```python
import matplotlib.pyplot as plt

price_counts.plot(kind='bar')

plt.title('Price Range Distribution')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')

plt.savefig('level1_task3_price_range.png')

plt.show()
```

```python
# Task 4: Online Delivery Analysis                                #level1:task4

delivery_counts = df['Has Online delivery'].value_counts()

print("Online Delivery Distribution:")
print(delivery_counts)

# Percentage
delivery_percentage = (delivery_counts / len(df)) * 100

print("\nPercentage Distribution:")
print(delivery_percentage.round(2))

# Average rating comparison
delivery_ratings = df.groupby('Has Online delivery')['Aggregate rating'].mean()

print("\nAverage Ratings:")
print(delivery_ratings)
```

```python
import matplotlib.pyplot as plt

delivery_counts.plot(kind='bar')

plt.title('Online Delivery Distribution')
plt.xlabel('Online Delivery')
plt.ylabel('Number of Restaurants')

plt.savefig('level1_task4_online_delivery.png')

plt.show()
```

```python
#restaurant chains
import pandas as pd                            #level2:task4
import matplotlib.pyplot as plt

# -----------------------------
# 1. Find restaurant chains
# -----------------------------
chain_counts = df['Restaurant Name'].value_counts()

chains = chain_counts[chain_counts > 1]

print("Restaurant Chains Found:")
print(chains.head(10))

# -----------------------------
# 2. Top chains (popularity)
# -----------------------------
top_chains = chain_counts.head(10)

print("\nTop Restaurant Chains (Popularity):")
print(top_chains)

# -----------------------------
# 3. Average rating of chains
# -----------------------------
chain_ratings = df[df['Restaurant Name'].isin(top_chains.index)] \
    .groupby('Restaurant Name')['Aggregate rating'] \
    .mean().sort_values(ascending=False)

print("\nAverage Ratings of Chains:")
print(chain_ratings)
```

```python
plt.figure(figsize=(10,5))

top_chains.plot(kind='bar')

plt.title("Top Restaurant Chains (Number of Outlets)")
plt.xlabel("Restaurant Name")
plt.ylabel("Number of Branches")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("level2_task4_chain_popularity.png")
plt.show()
```

```python
plt.figure(figsize=(10,5))

chain_ratings.plot(kind='bar', color='orange')

plt.title("Average Ratings of Restaurant Chains")
plt.xlabel("Restaurant Name")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("level2_task4_chain_ratings.png")
plt.show()
```

```python
import pandas as pd         #level3:tsak1

# =========================
# LEVEL 3 TASK 1
# RESTAURANT REVIEWS (SAFE VERSION)
# =========================

# 1. CHECK DATA
print(df.columns)

# 2. USE VOTES AS PROXY FOR ENGAGEMENT
df['review_proxy'] = df['Votes']

# 3. AVERAGE VALUE (PROXY)
avg_value = df['review_proxy'].mean()
print("Average Review Proxy (Votes):", avg_value)

# 4. CORRELATION BETWEEN VOTES AND RATING
correlation = df['review_proxy'].corr(df['Aggregate rating'])
print("Correlation (Votes vs Rating):", correlation)

# 5. POSITIVE & NEGATIVE CLASSIFICATION
positive = df[df['Aggregate rating'] >= 4.0]
negative = df[df['Aggregate rating'] <= 2.5]

print("Positive Restaurants:", len(positive))
print("Negative Restaurants:", len(negative))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(df['Votes'], df['Aggregate rating'], alpha=0.5)

plt.title("Votes vs Rating Analysis")
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")

plt.savefig("level3_task1votes_vs_rating.png")
plt.show()
```

```python
import pandas as pd                            #level3:task2

# =========================
# VOTES ANALYSIS
# =========================

# 1. HIGHEST VOTES RESTAURANTS
top_votes = df[['Restaurant Name', 'Votes']].sort_values(by='Votes', ascending=False).head(10)
print("Top 10 Restaurants by Votes:\n", top_votes)

# 2. LOWEST VOTES RESTAURANTS
low_votes = df[['Restaurant Name', 'Votes']].sort_values(by='Votes', ascending=True).head(10)
print("\nLowest 10 Restaurants by Votes:\n", low_votes)

# 3. CORRELATION BETWEEN VOTES AND RATING
correlation = df['Votes'].corr(df['Aggregate rating'])
print("\nCorrelation between Votes and Rating:", correlation)
```
